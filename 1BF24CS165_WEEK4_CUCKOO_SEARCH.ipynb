```python
import numpy as np
import math

# Objective function
def fitness(theta):
    return 1000 - (theta - 30) ** 2


# Generate Levy Flight step
def levy_flight():
    beta = 1.5

    sigma = (
        math.gamma(1 + beta) * math.sin(math.pi * beta / 2) /
        (math.gamma((1 + beta) / 2) * beta *
         2 ** ((beta - 1) / 2))
    ) ** (1 / beta)

    u = np.random.normal(0, sigma)
    v = np.random.normal(0, 1)

    return u / (abs(v) ** (1 / beta) + 1e-12)


# Cuckoo Search Algorithm
def cuckoo_search():

    N = 5
    Pa = 0.25
    max_iterations = 5

    lower = 0
    upper = 60
    alpha = 5

    # Initialize nests randomly
    nests = np.random.uniform(lower, upper, N)

    # Evaluate initial fitness
    best = nests[np.argmax([fitness(x) for x in nests])]
    best_fitness = fitness(best)

    print("Initial Population:")
    for i in range(N):
        print("Nest", i + 1, "Angle:", round(nests[i], 2),
              "Fitness:", round(fitness(nests[i]), 2))

    # Main iterations
    for iteration in range(1, max_iterations + 1):

        for i in range(N):

            # Generate new solution using Levy flight
            new_solution = nests[i] + alpha * levy_flight() * (
                np.random.random() * 2 - 1
            )

            new_solution = np.clip(new_solution, lower, upper)

            # Randomly select host nest
            j = np.random.randint(N)

            # Replace if new solution is better
            if fitness(new_solution) > fitness(nests[j]):
                nests[j] = new_solution

        # Abandon worst nests
        count = max(1, math.ceil(Pa * N))

        worst = np.argsort([fitness(x) for x in nests])[:count]

        for j in worst:
            nests[j] = np.random.uniform(lower, upper)

        # Preserve best solution
        current_best = nests[np.argmax([fitness(x) for x in nests])]

        if fitness(current_best) > best_fitness:
            best = current_best
            best_fitness = fitness(best)

        print("\nIteration:", iteration)
        print("Best Angle:", round(best, 2))
        print("Best Fitness:", round(best_fitness, 2))

    # Final result
    print("\nFinal Result")
    print("Optimal Tilt Angle:", round(best, 2), "degrees")
    print("Maximum Fitness:", round(best_fitness, 2))


# Execute algorithm
np.random.seed(42)
cuckoo_search()
```
