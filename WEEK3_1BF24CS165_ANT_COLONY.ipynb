```python
import random

# -----------------------------------------
# 1. INITIALIZE DISTANCE MATRIX
# -----------------------------------------

# 4 cities: 0, 1, 2, 3

distance = [
    [0, 20, 40, 50],
    [20, 0, 50, 20],
    [40, 50, 0, 20],
    [50, 20, 20, 0]
]

# Initialize pheromone matrix
pheromone = [
    [1.0 for j in range(4)]
    for i in range(4)
]


# -----------------------------------------
# 2. INITIALIZE PARAMETERS
# -----------------------------------------

num_cities = 4
ants = 4
iterations = 50

alpha = 1
beta = 2
rho = 0.5
Q = 100

best_tour = None
best_length = float('inf')

random.seed(42)


# -----------------------------------------
# 3. ANT COLONY OPTIMIZATION
# -----------------------------------------

for iteration in range(iterations):

    all_tours = []

    for ant in range(ants):

        # Select random starting city
        start = random.randint(0, num_cities - 1)

        tour = [start]

        unvisited = list(range(num_cities))
        unvisited.remove(start)

        # Construct the tour
        while unvisited:

            current_city = tour[-1]

            probabilities = []

            # Calculate probability of each unvisited city
            for city in unvisited:

                p = (
                    (pheromone[current_city][city] ** alpha)
                    *
                    ((1 / distance[current_city][city]) ** beta)
                )

                probabilities.append(p)

            # Normalize probabilities
            total = sum(probabilities)

            probabilities = [
                p / total for p in probabilities
            ]

            # Select next city using probability
            next_city = random.choices(
                unvisited,
                weights=probabilities,
                k=1
            )[0]

            tour.append(next_city)

            unvisited.remove(next_city)

        # Add starting city to complete the tour
        tour.append(start)

        # Calculate total tour length
        tour_length = 0

        for i in range(len(tour) - 1):

            tour_length += distance[
                tour[i]
            ][
                tour[i + 1]
            ]

        all_tours.append((tour, tour_length))

        # Update best tour
        if tour_length < best_length:

            best_length = tour_length
            best_tour = tour.copy()

    # -------------------------------------
    # 4. PHEROMONE EVAPORATION
    # -------------------------------------

    for i in range(num_cities):

        for j in range(num_cities):

            pheromone[i][j] *= (1 - rho)

    # -------------------------------------
    # 5. PHEROMONE DEPOSIT
    # -------------------------------------

    for tour, tour_length in all_tours:

        deposit = Q / tour_length

        for i in range(len(tour) - 1):

            city1 = tour[i]
            city2 = tour[i + 1]

            pheromone[city1][city2] += deposit
            pheromone[city2][city1] += deposit


# -----------------------------------------
# 6. DISPLAY OUTPUT
# -----------------------------------------

print("ANT COLONY OPTIMIZATION - TSP")
print("--------------------------------")

print("Best Tour:", " -> ".join(map(str, best_tour)))

print("Shortest Distance:", best_length)

print("\nFinal Pheromone Matrix:")

for row in pheromone:
    print([round(value, 2) for value in row])
```
